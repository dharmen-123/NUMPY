## Execution Time of List & Array 

#### %timeit method is used to check the time of execution of the datatype

#### LIST

```python
j=1
%timeit [j**2 for i in range(5)]
```

```python
j=1
b=%timeit [j**2 for i in range(5)]
print("Python List time-",b)
print(type(b))
```

#### Numpy Array

```python
import numpy as np
```

```python
%timeit np.arange(5)**2
```

```python
a = %timeit np.arange(5)**2
print("Numpy array time-",a)
print(type(a))
```
