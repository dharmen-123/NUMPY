## ****** NUMPY ARRAY INDEXING ******

```python
import numpy as np 
```

```python
arr=np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr[:5]
```

```python
arr[1:10:2]
```

```python
arr[::]
```

```python
arr[3::2]
```
