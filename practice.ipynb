         Dimesions in Array

```python
# 1-D array
x=[1,2,3,4,5]
x=np.array(x)
print(x.ndim)
```

```python
# 2-D Array
y=[[1,2],[3,4]]
y=np.array(y)
print(y.ndim)
```

```python
# N-D Array 
n=np.array([1,2,3],ndmin=10)
print(n.ndim)
print(n)
```
